Q1

```python
def search(n,t):
  l,h=0, len(n)-1
  while l <=h:
    m=(l+h)//2
    if n[m]==t:
      return m
    if n[l]<=n[m]:
      if n[l]<=t<n[m]:
          h=m-1
      else:
        l=m+1
    else:
      if n[m]<t<=n[h]:
        l=m+1
      else:
        h=m-1
  return -1
```

```python
print(search([4,5,6,7,0,1,2],0))
print(search([4,5,6,7,0,1,2],3))
```

```python
from collections import Counter
def min(s,t):
  if not s or not t:
    return ""
  n=Counter(t)
  miss=len(t)
  left=0
  best_l=float('inf')
  best_s=0
  for right , char in enumerate(s):
    if n[char]>0:
      miss-=1
    n[char]-=1
    if miss ==0:
      while n[s[left]]<0:
        n[s[left]]+=1
        left+=1
      if right - left +1< best_l:
        best_l=right - left +1
        best_s=left
      n[s[left]]+=1
      miss+=1
      left+=1
  return "" if best_l ==float('inf') else s[best_s:best_s+best_l]
```

```python
print(min("ADOBECODEBANC","ABC"))
print(min("a","a"))
```

Q3

```python
def premute(s):
  res=[]
  def backtrack(path,remaining):
    if not remaining:
      res.append(path)
      return
    for i in range(len(remaining)):
      backtrack(path + remaining[i],remaining[:i]+remaining[i+1:])
  backtrack("",s)
  return res
```

```python
print(premute("abc"))
print(premute("ab"))
```
